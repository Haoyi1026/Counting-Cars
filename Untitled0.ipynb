```python
import cv2
from ultralytics import YOLO


video_path = "/content/sample_data/traffic.mp4"
model = YOLO("yolov8n.pt")


cap = cv2.VideoCapture(video_path)
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

out = cv2.VideoWriter(
    'output_cars.mp4',
    cv2.VideoWriter_fourcc(*'mp4v'),
    30,
    (frame_width, frame_height)
)


line_x = frame_width // 2
counted_ids = set()
next_id = 0
vehicle_positions = {}

count_left_to_right = 0
count_right_to_left = 0
tolerance = 50


while True:
    ret, frame = cap.read()
    if not ret:
        break

    results = model.predict(frame, conf=0.5, verbose=False)
    detections = []


    for result in results:
        boxes = result.boxes.xyxy.cpu().numpy()
        class_ids = result.boxes.cls.cpu().numpy()

        for i, cls_id in enumerate(class_ids):
            if int(cls_id) in [2, 5, 7]:
                x1, y1, x2, y2 = boxes[i]
                detections.append((x1, y1, x2, y2))

    updated_positions = {}


    for box in detections:
        x1, y1, x2, y2 = box
        cx = int((x1 + x2)/2)
        cy = int((y1 + y2)/2)


        matched_id = None
        for vid, (prev_cx, prev_cy) in vehicle_positions.items():
            if abs(cx - prev_cx) < tolerance and abs(cy - prev_cy) < tolerance:
                matched_id = vid
                break

        if matched_id is None:
            matched_id = next_id
            next_id += 1

        updated_positions[matched_id] = (cx, cy)


        if matched_id not in counted_ids:
            if x1 <= line_x <= x2:
                counted_ids.add(matched_id)

                if cx < line_x:
                    count_right_to_left += 1
                else:
                    count_left_to_right += 1


        cv2.rectangle(frame, (int(x1), int(y1)), (int(x2), int(y2)), (0, 255, 0), 2)
        cv2.putText(frame, f"ID:{matched_id}", (cx-10, cy-10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 0, 0), 2)

    vehicle_positions = updated_positions


    cv2.line(frame, (line_x, 0), (line_x, frame_height), (0, 0, 255), 2)


    cv2.putText(frame, f"R->L: {count_right_to_left}", (50, 50),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)
    cv2.putText(frame, f"L->R: {count_left_to_right}", (50, 100),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 0, 0), 2)


    out.write(frame)

cap.release()
out.release()
cv2.destroyAllWindows()

print(f"Right->Left: {count_right_to_left}")
print(f"Left->Right: {count_left_to_right}")
print("Processed video saved as output_cars.mp4")
```
